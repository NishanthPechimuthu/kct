```python
# -------------------------------
# Task 1: Input Data
# -------------------------------
keys = ['A', 'B', 'C', 'D']

# Successful search probabilities p1..pn
p = [0, 0.15, 0.10, 0.05, 0.10]   # p[0] is dummy

# Unsuccessful search probabilities q0..qn
q = [0.05, 0.10, 0.05, 0.05, 0.10]

n = len(keys)


# -------------------------------
# Task 2, 3, 4: OBST DP
# -------------------------------
def optimal_bst(keys, p, q, n):
    # Create tables
    w = [[0]*(n+1) for _ in range(n+1)]   # weight table
    c = [[0]*(n+1) for _ in range(n+1)]   # cost table
    r = [[0]*(n+1) for _ in range(n+1)]   # root table

    # Base cases
    for i in range(n+1):
        w[i][i] = q[i]
        c[i][i] = 0
        r[i][i] = 0

    # DP for subtree lengths
    for length in range(1, n+1):
        for i in range(0, n-length+1):
            j = i + length

            # Compute weight
            w[i][j] = w[i][j-1] + p[j] + q[j]

            # Set initial cost to infinity
            c[i][j] = float('inf')

            # Try every key as root
            for k in range(i+1, j+1):
                cost = c[i][k-1] + c[k][j] + w[i][j]

                if cost < c[i][j]:
                    c[i][j] = cost
                    r[i][j] = k

    return w, c, r


# -------------------------------
# Task 5: Print OBST Structure
# -------------------------------
def print_obst(r, keys, i, j, parent=None, side="root"):
    if i == j:
        return

    root_index = r[i][j]
    current_key = keys[root_index - 1]

    if parent is None:
        print(f"{current_key} is the root")
    else:
        print(f"{current_key} is the {side} child of {parent}")

    # Left subtree
    print_obst(r, keys, i, root_index - 1, current_key, "left")

    # Right subtree
    print_obst(r, keys, root_index, j, current_key, "right")


# -------------------------------
# MAIN
# -------------------------------
w, c, r = optimal_bst(keys, p, q, n)

print("Minimum Expected Search Cost =", round(c[0][n], 2))
print("\nOptimal Binary Search Tree Structure:")
print_obst(r, keys, 0, n)
```
