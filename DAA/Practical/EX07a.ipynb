```python
class Item:
    def __init__(self, id, value, weight):
        self.id = id
        self.value = value
        self.weight = weight
        self.ratio = value / weight

def fractional_knapsack(items, capacity):
    items = sorted(items, key=lambda item: item.ratio, reverse=True)

    total_value = 0
    steps = []

    for item in items:
        if capacity <= 0:
            break

        if item.weight <= capacity:
            taken_weight = item.weight
            taken_value = item.value
            fraction = 1
        else:
            taken_weight = capacity
            fraction = capacity / item.weight
            taken_value = item.value * fraction

        total_value += taken_value
        capacity -= taken_weight

        steps.append((item.id, taken_weight, taken_value, fraction, capacity))

    return total_value, steps


def print_steps(steps, total_value):
    print("\nItem\tTaken Wt\tAdded Value\tFraction\tRemaining")
    print("-" * 65)

    for item_id, wt, val, frac, rem in steps:
        print(f"{item_id}\t{wt:.2f}\t\t{val:.2f}\t\t{frac:.2f}\t\t{rem:.2f}")

    print("-" * 65)
    print(f"Maximum Value = {total_value:.2f}")


# Example
items = [
    Item(1, 60, 10),
    Item(2, 100, 20),
    Item(3, 120, 30)
]

capacity = 50

total_value, steps = fractional_knapsack(items, capacity)
print_steps(steps, total_value)
```
