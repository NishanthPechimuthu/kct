```python
# -------------------------------
# Task 1: Recursive LCS
# -------------------------------
def lcs_recursive(X, Y, m, n):
    if m == 0 or n == 0:
        return 0

    if X[m-1] == Y[n-1]:
        return 1 + lcs_recursive(X, Y, m-1, n-1)

    return max(lcs_recursive(X, Y, m-1, n),
               lcs_recursive(X, Y, m, n-1))


# -------------------------------
# Task 2: DP (2D Table)
# -------------------------------
def lcs_dp(X, Y):
    m, n = len(X), len(Y)
    L = [[0]*(n+1) for _ in range(m+1)]

    for i in range(1, m+1):
        for j in range(1, n+1):
            if X[i-1] == Y[j-1]:
                L[i][j] = 1 + L[i-1][j-1]
            else:
                L[i][j] = max(L[i-1][j], L[i][j-1])

    return L


# -------------------------------
# Task 3: Direction Matrix
# -------------------------------
def lcs_with_direction(X, Y):
    m, n = len(X), len(Y)

    L = [[0]*(n+1) for _ in range(m+1)]
    D = [[""]*(n+1) for _ in range(m+1)]

    for i in range(1, m+1):
        for j in range(1, n+1):
            if X[i-1] == Y[j-1]:
                L[i][j] = 1 + L[i-1][j-1]
                D[i][j] = "D"   # Diagonal
            elif L[i-1][j] >= L[i][j-1]:
                L[i][j] = L[i-1][j]
                D[i][j] = "U"   # Up
            else:
                L[i][j] = L[i][j-1]
                D[i][j] = "L"   # Left

    return L, D


# -------------------------------
# Task 4: Backtracking (Print LCS)
# -------------------------------
def print_lcs(D, X, i, j):
    if i == 0 or j == 0:
        return ""

    if D[i][j] == "D":
        return print_lcs(D, X, i-1, j-1) + X[i-1]
    elif D[i][j] == "U":
        return print_lcs(D, X, i-1, j)
    else:
        return print_lcs(D, X, i, j-1)


# -------------------------------
# Task 5: Space Optimized
# -------------------------------
def lcs_space_optimized(X, Y):
    if len(X) < len(Y):
        X, Y = Y, X

    prev = [0]*(len(Y)+1)
    curr = [0]*(len(Y)+1)

    for i in range(1, len(X)+1):
        for j in range(1, len(Y)+1):
            if X[i-1] == Y[j-1]:
                curr[j] = 1 + prev[j-1]
            else:
                curr[j] = max(prev[j], curr[j-1])

        prev = curr[:]

    return prev[len(Y)]


# -------------------------------
# MAIN TEST
# -------------------------------
X = "AGGTABCDXY"
Y = "GXTXAYBCDZ"

print("Recursive LCS Length:",
      lcs_recursive(X, Y, len(X), len(Y)))

L = lcs_dp(X, Y)
print("DP LCS Length:", L[len(X)][len(Y)])

L, D = lcs_with_direction(X, Y)
lcs_str = print_lcs(D, X, len(X), len(Y))

print("LCS String:", lcs_str)

print("Space Optimized LCS Length:",
      lcs_space_optimized(X, Y))
```
