```python
# -------------------------------
# Task 1: Graph using Adjacency List
# -------------------------------
class Graph:
    def __init__(self):
        self.adj = {}
        self.edges = {}

    def add_edge(self, u, v):
        if u not in self.adj:
            self.adj[u] = []
        if v not in self.adj:
            self.adj[v] = []

        self.adj[u].append(v)
        self.adj[v].append(u)

        edge = tuple(sorted((u, v)))
        self.edges[edge] = False   # False = unvisited

    def degree(self):
        return {node: len(neighbors) for node, neighbors in self.adj.items()}

    def get_unvisited_edges(self):
        return [e for e, visited in self.edges.items() if not visited]

    def mark_visited(self, u, v):
        edge = tuple(sorted((u, v)))
        if edge in self.edges:
            self.edges[edge] = True


# -------------------------------
# Task 2: Greedy Approximation
# -------------------------------
def vertex_cover_approx(graph):
    cover = set()

    while True:
        unvisited = graph.get_unvisited_edges()
        if not unvisited:
            break

        # Pick any unvisited edge
        u, v = unvisited[0]

        # Add both endpoints to cover
        cover.add(u)
        cover.add(v)

        # Mark all incident edges of u and v as visited
        for neighbor in graph.adj[u]:
            graph.mark_visited(u, neighbor)

        for neighbor in graph.adj[v]:
            graph.mark_visited(v, neighbor)

    return cover


# -------------------------------
# Utility to print result
# -------------------------------
def test_graph(name, graph, optimal_size):
    print(name)
    print("Degrees:", graph.degree())

    approx_cover = vertex_cover_approx(graph)
    print("Approx Vertex Cover:", approx_cover)
    print("Approx Cover Size  :", len(approx_cover))
    print("Optimal Cover Size :", optimal_size)
    print()


# -------------------------------
# Task 3: Test Cases
# -------------------------------

# 1) Star Graph
g1 = Graph()
for i in range(1, 6):
    g1.add_edge(0, i)

# 2) Cycle Graph C4
g2 = Graph()
cycle_edges = [(0,1), (1,2), (2,3), (3,0)]
for u, v in cycle_edges:
    g2.add_edge(u, v)

# 3) Complete Graph K4
g3 = Graph()
nodes = [0,1,2,3]
for i in range(len(nodes)):
    for j in range(i+1, len(nodes)):
        g3.add_edge(nodes[i], nodes[j])


# Run tests
test_graph("Star Graph", g1, optimal_size=1)
test_graph("Cycle Graph C4", g2, optimal_size=2)
test_graph("Complete Graph K4", g3, optimal_size=3)
```
