```python
# -------------------------------
# Task 1: Check if safe to place queen
# -------------------------------
def isSafe(board, row, col, n):

    # Check same row (left side only)
    for i in range(col):
        if board[row][i] == 1:
            return False

    # Upper diagonal
    i, j = row, col
    while i >= 0 and j >= 0:
        if board[i][j] == 1:
            return False
        i -= 1
        j -= 1

    # Lower diagonal
    i, j = row, col
    while i < n and j >= 0:
        if board[i][j] == 1:
            return False
        i += 1
        j -= 1

    return True


# -------------------------------
# Task 2 & 3: Backtracking
# -------------------------------
def solve_util(board, col, n, solutions):

    # If all queens placed
    if col == n:
        solutions.append([row[:] for row in board])
        return

    for row in range(n):
        if isSafe(board, row, col, n):
            board[row][col] = 1

            solve_util(board, col + 1, n, solutions)

            # Backtrack
            board[row][col] = 0


def solve_n_queens(n, find_all=False):
    board = [[0]*n for _ in range(n)]
    solutions = []

    solve_util(board, 0, n, solutions)

    if not find_all and solutions:
        return [solutions[0]]  # first solution only

    return solutions


# -------------------------------
# Print board
# -------------------------------
def print_board(board):
    for row in board:
        print(" ".join("Q" if x else "." for x in row))
    print()


# -------------------------------
# MAIN
# -------------------------------
n = 4

print("Task 2: First solution")
first = solve_n_queens(n, find_all=False)
if first:
    print_board(first[0])
else:
    print("No solution")

print("Task 3: All solutions")
all_solutions = solve_n_queens(n, find_all=True)

print("Total solutions =", len(all_solutions), "\n")

for i, sol in enumerate(all_solutions, 1):
    print(f"Solution {i}:")
    print_board(sol)
```
