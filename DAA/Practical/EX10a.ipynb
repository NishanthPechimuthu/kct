```python
def isSafe(board, row, col, n):

    for i in range(col):
        if board[row][i] == 1:
            return False
    
    for i, j in zip(range(row, -1, -1), range(col, -1, -1)):
        if board[i][j] == 1:
            return False
    
    for i, j in zip(range(row, n, 1), range(col, -1, -1)):
        if board[i][j] == 1:
            return False
    
    return True
```

```python
def solveNQueensUtil(board, col, n, solutions):

    if col == n:
        solutions.append([row[:] for row in board])
        return
    
    for row in range(n):
        if isSafe(board, row, col, n):
            board[row][col] = 1
            
            solveNQueensUtil(board, col + 1, n, solutions)
            
            board[row][col] = 0
```

```python
def solveNQueens(n, find_all=False):

    board = [[0 for _ in range(n)] for _ in range(n)]
    solutions = []
    
    solveNQueensUtil(board, 0, n, solutions)
    
    if not find_all and solutions:
        return [solutions[0]]
    return solutions
```

```python
def printBoard(board):
    for row in board:
        print(" ".join("Q" if cell == 1 else "." for cell in row))
    print()
```

```python
def main():
    print("Task 2: First solution for N = 4")
    sol = solveNQueens(4, find_all=False)
    if sol:
        printBoard(sol[0])
    else:
        print("No solution")
```

```python
def main():
    print("Task 2: First solution for N = 4")
    sol = solveNQueens(4, find_all=False)
    if sol:
        printBoard(sol[0])
    else:
        print("No solution")

    print("\nTask 3: All solutions and count for N = 4")
    all_solutions = solveNQueens(4, find_all=True)
    print(f"Total number of distinct solutions = {len(all_solutions)}\n")
    
    for i, board in enumerate(all_solutions, 1):
        print(f"Solution #{i}:")
        printBoard(board)
```

```python
if __name__ == "__main__":
    main()
```
