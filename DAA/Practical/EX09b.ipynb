```python
INF = 9999

# -------------------------------
# Task 1: Graph as Adjacency Matrix
# -------------------------------
graph = [
    [0,    3,    INF,  7],
    [8,    0,    2,    INF],
    [5,    INF,  0,    1],
    [2,    INF,  INF,  0]
]

n = len(graph)


# -------------------------------
# Task 2, 3, 5: Floyd-Warshall
# -------------------------------
def floyd_warshall(graph):
    n = len(graph)

    # Distance matrix
    dist = [row[:] for row in graph]

    # Path / predecessor-like matrix
    nxt = [[None]*n for _ in range(n)]

    # Initialize path matrix
    for i in range(n):
        for j in range(n):
            if graph[i][j] != INF and i != j:
                nxt[i][j] = j

    # Core Floyd-Warshall
    for k in range(n):
        for i in range(n):
            for j in range(n):
                if dist[i][k] != INF and dist[k][j] != INF:
                    if dist[i][j] > dist[i][k] + dist[k][j]:
                        dist[i][j] = dist[i][k] + dist[k][j]
                        nxt[i][j] = nxt[i][k]

    # Task 5: Negative cycle check
    has_negative_cycle = False
    for i in range(n):
        if dist[i][i] < 0:
            has_negative_cycle = True
            break

    return dist, nxt, has_negative_cycle


# -------------------------------
# Task 4: Print Path
# -------------------------------
def print_path(nxt, i, j):
    if nxt[i][j] is None:
        print(f"No path from {i} to {j}")
        return

    path = [i]
    while i != j:
        i = nxt[i][j]
        path.append(i)

    print(" -> ".join(map(str, path)))


# -------------------------------
# MAIN
# -------------------------------
dist, nxt, has_negative_cycle = floyd_warshall(graph)

print("Shortest Distance Matrix:")
for row in dist:
    print(row)

print("\nShortest Paths:")
print("Path 0 to 3:", end=" ")
print_path(nxt, 0, 3)

print("Path 3 to 1:", end=" ")
print_path(nxt, 3, 1)

print("\nNegative Cycle Check:")
if has_negative_cycle:
    print("Graph contains a NEGATIVE WEIGHT CYCLE.")
else:
    print("No negative weight cycle detected.")
```
