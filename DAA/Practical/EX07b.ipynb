```python
class Activity:
    def __init__(self, id, start, finish):
        self.id = id
        self.start = start
        self.finish = finish
        self.duration = finish - start


# Task 1: Correct Greedy Strategy (Sort by Finish Time)
def activity_selection_finish_time(activities):
    activities = sorted(activities, key=lambda x: x.finish)

    selected = []
    last_finish = float('-inf')

    for act in activities:
        if act.start >= last_finish:
            selected.append(act)
            last_finish = act.finish

    return selected


# Task 3: Wrong Strategy (Sort by Shortest Duration)
def activity_selection_shortest_duration(activities):
    activities = sorted(activities, key=lambda x: x.duration)

    selected = []
    last_finish = float('-inf')

    for act in activities:
        if act.start >= last_finish:
            selected.append(act)
            last_finish = act.finish

    return selected


def print_selected(title, selected):
    print(f"\n{title}")
    print("-" * 45)

    if not selected:
        print("No activities selected.")
        return

    ids = [act.id for act in selected]
    print("Selected Activity IDs :", ids)
    print("Maximum Activities    :", len(selected))
    print("Activities Chosen:")

    for act in selected:
        print(f"A{act.id}: {act.start} -> {act.finish}")


# -------------------------------
# Task 1: Sorting and Selection
# -------------------------------
activities1 = [
    Activity(1, 1, 4),
    Activity(2, 3, 5),
    Activity(3, 0, 6),
    Activity(4, 5, 7),
    Activity(5, 3, 9),
    Activity(6, 5, 9),
    Activity(7, 6, 10),
    Activity(8, 8, 11),
    Activity(9, 8, 12),
    Activity(10, 2, 14)
]

print("=== Task 1: Sorting by Finish Time and Greedy Selection ===")
selected1 = activity_selection_finish_time(activities1)
print_selected("Optimal Selection (Finish Time)", selected1)


# --------------------------------------
# Task 2: User-Defined Input Testing
# Example: 5 activities where overlaps exist
# --------------------------------------
activities2 = [
    Activity(1, 1, 10),
    Activity(2, 2, 6),
    Activity(3, 5, 9),
    Activity(4, 7, 11),
    Activity(5, 10, 12)
]

print("\n=== Task 2: User-Defined Overlapping Activities ===")
selected2 = activity_selection_finish_time(activities2)
print_selected("Selection for User Test Case", selected2)


# --------------------------------------
# Task 3: Strategy Counter-Proof
# Show shortest duration is not always optimal
# --------------------------------------
activities3 = [
    Activity(1, 1, 10),
    Activity(2, 2, 6),
    Activity(3, 5, 9),
    Activity(4, 7, 11),
    Activity(5, 10, 12)
]

print("\n=== Task 3: Counter-Proof for Shortest Duration Strategy ===")

selected_finish = activity_selection_finish_time(activities3)
selected_duration = activity_selection_shortest_duration(activities3)

print_selected("Correct Strategy: Sort by Finish Time", selected_finish)
print_selected("Wrong Strategy: Sort by Shortest Duration", selected_duration)

print("\nConclusion:")
print("Sorting by Finish Time gives the optimal maximum set of activities.")
print("Sorting by Shortest Duration can lead to a sub-optimal solution.")
```
