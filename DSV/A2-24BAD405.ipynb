**Rolling Sensor Analysis**

```python
nr() 

sensor = np.array([22.1, 22.5, 23.0, 24.8, 25.1, 26.0, 27.5, 28.0])

rolling_mean = pd.Series(sensor).rolling(window=3).mean()
rolling_std = pd.Series(sensor).rolling(window=3).std()

threshold = 1.5 * rolling_std
anomalies = sensor[(sensor > rolling_mean + threshold) | 
                   (sensor < rolling_mean - threshold)]

print("Rolling Mean:\n", rolling_mean)
print("\nRolling Std Dev:\n", rolling_std)
print("\nAnomalous Values:\n", anomalies)
```

**Student Performance Analysis**

```python
nr()
data = {
    "Student": ["A", "B", "C", "D", "E"],
    "Math": [78, 85, 67, 90, 72],
    "Physics": [74, 88, 69, 92, 70],
    "Attendance": [82, 90, 76, 95, 80]
}

df = pd.DataFrame(data)
df["Average"] = df[["Math", "Physics", "Attendance"]].mean(axis=1)

def grade(avg):
    if avg >= 85: return "A"
    elif avg >= 70: return "B"
    else: return "C"

df["Grade"] = df["Average"].apply(grade)
top_student = df.loc[df["Average"].idxmax()]

print(df)
print("\nTop Performer:\n", top_student)
```

**Time-Series Feature Engineering**

```python
nr()

timestamps = pd.date_range("2024-01-01", periods=10, freq="6H")
values = [12, 18, 20, 22, 19, 15, 14, 17, 21, 23]

ts_df = pd.DataFrame({"Timestamp": timestamps, "Value": values})
ts_df["Hour"] = ts_df["Timestamp"].dt.hour
ts_df["Day"] = ts_df["Timestamp"].dt.day

ts_df["Hour_sin"] = np.sin(2 * np.pi * ts_df["Hour"] / 24)
ts_df["Hour_cos"] = np.cos(2 * np.pi * ts_df["Hour"] / 24)

peak_hour = ts_df.loc[ts_df["Value"].idxmax(), "Hour"]

print(ts_df)
print("\nPeak Activity Hour:", peak_hour)
```

**Data Leakage Detection**

```python
nr()
train = pd.DataFrame({
    "id": [1, 2, 3, 4],
    "score": [65, 70, 80, 90],
    "result": [0, 0, 1, 1]
})

test = pd.DataFrame({
    "id": [3, 5],
    "score": [80, 75],
    "result": [1, 0]
})

overlap = set(train["id"]).intersection(set(test["id"]))
leakage_percent = (len(overlap) / len(test)) * 100

print("Overlapping IDs:", overlap)
print("Potential Leakage Columns: id, result")
print("Leakage Percentage:", leakage_percent)
```

```python
nr()
df = pd.DataFrame({
    "A": np.random.randint(0, 1000, 10000),
    "B": np.random.rand(10000),
    "C": ["category"] * 10000
})

before = df.memory_usage(deep=True).sum()

df["A"] = df["A"].astype("int16")
df["B"] = df["B"].astype("float32")
df["C"] = df["C"].astype("category")

after = df.memory_usage(deep=True).sum()

print("Memory Before Optimization:", before)
print("Memory After Optimization:", after)
```

**Uncertainty Visualization**

```python
nr()
x = np.arange(1, 11)
y = np.array([10, 12, 15, 18, 20, 22, 24, 23, 25, 27])
error = np.array([1, 1.2, 1.5, 1.3, 1.4, 1.6, 1.8, 1.7, 1.9, 2.0])

max_err_idx = error.argmax()

plt.figure()
plt.plot(x, y)
plt.fill_between(x, y-error, y+error, alpha=0.3)
plt.scatter(x[max_err_idx], y[max_err_idx])
plt.annotate("Max Uncertainty", (x[max_err_idx], y[max_err_idx]))
plt.xlabel("X")
plt.ylabel("Y")
plt.title("Uncertainty Visualization")
plt.show()
```

**Correlation Heatmap**

```python
nr()
df = pd.DataFrame({
    "Temp": [30, 32, 34, 36, 38],
    "Humidity": [70, 65, 60, 55, 50],
    "Pressure": [1012, 1010, 1008, 1006, 1004]
})

corr = df.corr()

plt.imshow(corr, cmap="coolwarm")
plt.colorbar()
plt.xticks(range(len(corr.columns)), corr.columns)
plt.yticks(range(len(corr.columns)), corr.columns)

for i in range(len(corr)):
    for j in range(len(corr)):
        plt.text(j, i, f"{corr.iloc[i,j]:.2f}", ha="center", va="center")

plt.title("Correlation Heatmap")
plt.show()
```
